```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from datasets import load_dataset
from collections import Counter

# Настройка устройства
device = torch.device("mps" if torch.backends.mps.is_available() else 
                     "cuda" if torch.cuda.is_available() else "cpu")

# 1. Загрузка и подготовка данных
dataset = load_dataset("wikitext", "wikitext-2-raw-v1", split="train[:2%]")
texts = [t for t in dataset['text'] if t and len(t.split()) > 2][:10000]

# Создаем словарь
word_counts = Counter()
for text in texts:
    word_counts.update(text.lower().split())

vocab = {'<PAD>': 0, '<UNK>': 1, '<SOS>': 2, '<EOS>': 3}
for i, (word, _) in enumerate(word_counts.most_common(5000), 4):
    vocab[word] = i

vocab_size = len(vocab)
inv_vocab = {v: k for k, v in vocab.items()}

# 2. Простой датасет
class TextData(Dataset):
    def __init__(self, texts, vocab, seq_len=16):
        self.seq_len = seq_len
        self.data = []
        
        for text in texts:
            words = text.lower().split()[:seq_len+2]
            ids = [vocab.get(w, 1) for w in words]
            
            for i in range(len(ids)-seq_len):
                self.data.append(ids[i:i+seq_len+1])
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        seq = self.data[idx]
        return torch.tensor(seq[:-1]), torch.tensor(seq[1:])

# 3. Упрощенная модель
class SimpleTransformer(nn.Module):
    def __init__(self, vocab_size, d_model=128):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, 
            nhead=4,
            dim_feedforward=256,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=2)
        self.out = nn.Linear(d_model, vocab_size)
    
    def forward(self, x):
        x = self.embed(x)
        x = self.pos_encoder(x)
        mask = torch.triu(torch.full((x.size(1), x.size(1)), float('-inf')), diagonal=1)
        x = self.transformer(x, mask=mask.to(x.device))
        return self.out(x)

# Позиционное кодирование
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=100):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-torch.log(torch.tensor(10000.0)) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)
    
    def forward(self, x):
        return x + self.pe[:, :x.size(1)]

# 4. Функция генерации
def generate(model, start, vocab, inv_vocab, steps=10):
    model.eval()
    words = start.lower().split()
    ids = [vocab.get(w, 1) for w in words]
    
    with torch.no_grad():
        for _ in range(steps):
            inp = torch.tensor([ids[-16:]]).to(device)
            out = model(inp)
            probs = torch.softmax(out[0, -1], dim=-1)
            next_id = torch.multinomial(probs, 1).item()
            
            if next_id == vocab['<EOS>'] or len(words) > 25:
                break
                
            ids.append(next_id)
            words.append(inv_vocab.get(next_id, '?'))
    
    return ' '.join(words)

# 5. Обучение
model = SimpleTransformer(vocab_size).to(device)
dataset = TextData(texts[:800], vocab)
val_data = TextData(texts[800:], vocab)

loader = DataLoader(dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_data, batch_size=32)

opt = torch.optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

for epoch in range(15):
    model.train()
    total_loss = 0
    
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        out = model(x)
        loss = criterion(out.view(-1, vocab_size), y.view(-1))
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
        total_loss += loss.item()
    
    print(f"Эпоха {epoch+1}: Loss: {total_loss/len(loader):.3f}")
    
    # Генерация примеров
    if epoch % 2 == 0:
        print("Примеры:")
        for start in ["the cat", "computer"]:
            result = generate(model, start, vocab, inv_vocab)
            print(f"  '{start}' -> '{result}'")

# 6. Тест
print("\nФинальные тесты:")
test_inputs = ["the cat", "machine learning"]
for inp in test_inputs:
    result = generate(model, inp, vocab, inv_vocab, steps=12)
    print(f"'{inp}' -> '{result}'")
```
