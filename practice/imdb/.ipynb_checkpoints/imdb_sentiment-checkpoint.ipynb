# Классификация тональности 

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from datasets import load_dataset
import re
from collections import Counter
import numpy as np
from tqdm import tqdm

DEVICE = torch.device('mps' if torch.backends.mps.is_available() else 'cpu')
print(f"Устройство: {DEVICE}\n")
```

```python
print("📥 Загрузка данных IMDB...")
ds = load_dataset("imdb")
```

```python
import os
from pathlib import Path
from datasets import config

def get_size(path):
    """Возвращает размер папки/файла в байтах"""
    path = Path(path)
    if path.is_file():
        return path.stat().st_size
    total = 0
    for entry in path.rglob('*'):
        if entry.is_file():
            total += entry.stat().st_size
    return total

# Кэш, где хранятся все датасеты из библиотеки datasets
cache_dir = Path(config.HF_DATASETS_CACHE)
print(f"Кэш датасетов: {cache_dir}\n")

total_size = get_size(cache_dir)
print(f"Общий размер кэша: {total_size / (1024**2):.2f} MB")

# Размер конкретного датасета (например, IMDB)
imdb_dirs = list(cache_dir.glob("imdb*"))
for d in imdb_dirs:
    size = get_size(d)
    print(f"\nДатасет: {d.name}")
    print(f"Размер: {size / (1024**2):.2f} MB ({size / (1024**3):.2f} GB)")
    print(f"Путь: {d}")
```

# Токенизируем текст 

```python
def tokenize(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    return text.split()
```

```python
print("Создание словаря...")
counter = Counter()
for ex in tqdm(ds["train"], desc="Подсчёт токенов"):
    counter.update(tokenize(ex["text"]))
```

```python
print(dict(list(counter.items())[:15]))
```

# Создаем словарь

```python
specials = {'<pad>': 0, '<unk>': 1, '<bos>': 2, '<eos>': 3}
vocab = {word: idx + len(specials) for idx, (word, _) in enumerate(counter.most_common(10000))}
vocab = {**specials, **vocab}
PAD_IDX = vocab['<pad>']
UNK_IDX = vocab['<unk>']
print(f"Словарь готов: {len(vocab)} токенов\n")
```

```python
print(dict(list(vocab.items())[:15]))
```

# Создаем класс IMDBDataset

```python
class IMDBDataset(Dataset):
    def __init__(self, data, max_len=200):
        self.data, self.labels = [], []
        for ex in tqdm(data, desc="Подготовка данных", leave=False):
            tokens = tokenize(ex["text"])
            ids = [vocab.get('<bos>', 2)] + [vocab.get(t, UNK_IDX) for t in tokens[:max_len-2]] + [vocab.get('<eos>', 3)]
            ids += [PAD_IDX] * (max_len - len(ids))
            self.data.append(ids[:max_len])
            self.labels.append(ex["label"])
    
    def __len__(self): return len(self.data)
    def __getitem__(self, i): 
        return torch.tensor(self.data[i], dtype=torch.long), torch.tensor(self.labels[i], dtype=torch.long)

print("Создание датасетов...")
train_ds = IMDBDataset(ds["train"])
test_ds = IMDBDataset(ds["test"])
train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=32)
print(f"Train: {len(train_ds)} примеров, Test: {len(test_ds)} примеров\n")
```

# Positional encoding

```python
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(max_len).unsqueeze(1).float()
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        self.register_buffer('pe', pe.unsqueeze(0))
    
    def forward(self, x): return x + self.pe[:, :x.size(1)]
```

# Создаем модель 

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(len(vocab), 128, padding_idx=PAD_IDX)
        self.pos = PositionalEncoding(128)
        layer = nn.TransformerEncoderLayer(128, nhead=4, batch_first=True, dim_feedforward=512)
        self.encoder = nn.TransformerEncoder(layer, num_layers=2)
        self.classifier = nn.Sequential(nn.Dropout(0.1), nn.Linear(128, 2))
    
    def forward(self, x):
        x = self.pos(self.embed(x))
        x = self.encoder(x) 
        return self.classifier(x[:, 0])
```

# Обучение

```python
model = Model().to(DEVICE)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
num_of_epochs = 3

print("Начало обучения")
for epoch in range(num_of_epochs):
    model.train()
    train_loss = 0
    progress = tqdm(train_loader, desc=f"Эпоха {epoch+1}/3", leave=False)
    for data, target in progress:
        data, target = data.to(DEVICE), target.to(DEVICE)
        optimizer.zero_grad()
        loss = criterion(model(data), target)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
        progress.set_postfix({'loss': f'{loss.item():.4f}'})
    
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for data, target in tqdm(test_loader, desc="Тестирование", leave=False):
            data, target = data.to(DEVICE), target.to(DEVICE)
            preds = model(data).argmax(dim=1)
            correct += (preds == target).sum().item()
            total += target.size(0)
    
    acc = correct / total
    print(f"Эпоха {epoch+1}/3 | Loss: {train_loss/len(train_loader):.4f} | Точность: {acc:.2%}")
```

# Функция для предсказаний

```python
def predict(text):
    tokens = tokenize(text)
    ids = [vocab.get('<bos>', 2)] + [vocab.get(t, UNK_IDX) for t in tokens[:198]] + [vocab.get('<eos>', 3)]
    ids += [PAD_IDX] * (200 - len(ids))
    x = torch.tensor(ids[:200], dtype=torch.long).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        pred = model(x).argmax().item()
    return "👍 Положительный" if pred == 1 else "👎 Отрицательный"
```
