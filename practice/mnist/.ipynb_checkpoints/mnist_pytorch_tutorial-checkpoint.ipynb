# Нейросеть на PyTorch за 10 клеток — распознавание цифр (MNIST)

Это учебный ноутбук для **новичков**. Мы напишем и обучим простую нейросеть,
которая по картинке рукописной цифры (0–9) угадывает, какая это цифра.

**Датасет MNIST** — это 70 000 чёрно-белых картинок размером 28×28 пикселей.
Классика для первого знакомства с нейросетями.

Что будет по шагам:

2. Выбор устройства (CPU / GPU)
3. Загрузка данных
4. Смотрим, что внутри данных (вход и выход)
5. Рисуем несколько примеров
6. Описываем саму нейросеть
7. Функция потерь и оптимизатор
8. Обучение (тренировочный цикл)
9. Проверка точности на тестовых данных
10. Демонстрация: даём сети картинку и смотрим предсказание

## Шаг 1. Импорт библиотек

- `torch` — сам PyTorch (тензоры, автоматическое дифференцирование).
- `torch.nn` — «кирпичики» для построения нейросетей (слои, функции потерь).
- `torchvision` — готовые датасеты (в т.ч. MNIST) и преобразования картинок.
- `matplotlib` — чтобы рисовать картинки.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

print("Версия PyTorch:", torch.__version__)
```

## Шаг 2. Выбор устройства

Нейросети считаются быстрее на видеокарте (GPU). Если её нет — используем
процессор (CPU). Код одинаковый: PyTorch сам перенесёт вычисления туда,
куда мы скажем.


```python
device = torch.device("mps" if torch.mps.is_available() else "cpu")
print("Обучение будет на:", device)
```

## Шаг 3. Загрузка данных MNIST

`transforms.ToTensor()` делает две вещи:
1. Превращает картинку в **тензор** (многомерный массив чисел).
2. Нормализует яркость пикселей из диапазона 0–255 в диапазон **0.0–1.0**.

`train=True` — обучающая выборка (60 000 картинок),
`train=False` — тестовая выборка (10 000 картинок, на них проверяем качество).

При первом запуске данные скачаются в папку `./data` (несколько мегабайт).


```python
transform = transforms.ToTensor()

train_dataset = datasets.MNIST(root="./data", train=True,  download=True, transform=transform)
test_dataset  = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

print("Картинок для обучения:", len(train_dataset))
print("Картинок для теста:   ", len(test_dataset))
```

## Шаг 4. Что внутри данных? (вход и выход)

Каждый элемент датасета — это пара `(картинка, метка)`:
- **картинка** — тензор формы `[1, 28, 28]`
  (1 канал — чёрно-белая; 28×28 — высота и ширина в пикселях).
- **метка (label)** — просто число от 0 до 9: какая это цифра.

Это и есть наши **вход** (картинка) и **правильный ответ** (метка),
которому мы учим сеть.


```python
image, label = train_dataset[0]   # берём самый первый пример

print("Форма (shape) картинки:", image.shape)     # torch.Size([1, 28, 28])
print("Тип данных картинки:    ", image.dtype)     # float32
print("Мин/макс яркость:        %.2f / %.2f" % (image.min(), image.max()))
print("Правильный ответ (метка):", label)          # например, 5
```

## Шаг 5. Смотрим на примеры глазами

Нарисуем первые 8 картинок из обучающей выборки вместе с их метками,
чтобы понимать, с чем работаем.


```python
fig, axes = plt.subplots(1, 8, figsize=(12, 2))

for i in range(8):
    img, lbl = train_dataset[i]
    axes[i].imshow(img.squeeze(), cmap="gray")  # squeeze убирает лишнюю размерность [1,28,28] -> [28,28]
    axes[i].set_title(f"Метка: {lbl}")
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

## Шаг 6. DataLoader — подаём данные пачками (батчами)

Обучать по одной картинке медленно. `DataLoader` собирает данные в **батчи**
(пачки, например по 64 картинки) и по желанию перемешивает их (`shuffle=True`).

- Обучающие данные перемешиваем — так сеть учится лучше.
- Тестовые перемешивать не нужно.


```python
batch_size = 64

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)

# Посмотрим на форму одного батча:
images, labels = next(iter(train_loader))
print("Форма батча картинок:", images.shape)  # [64, 1, 28, 28] -> 64 картинки
print("Форма батча меток:   ", labels.shape)  # [64]
```

## Шаг 7. Описываем нейросеть

Сделаем самую простую **полносвязную** сеть (её ещё зовут MLP):

1. `Flatten` — «разворачивает» картинку 28×28 в один длинный вектор из 784 чисел.
2. `Linear(784, 128)` — скрытый слой из 128 нейронов.
3. `ReLU` — функция активации (оставляет положительные числа, обнуляет отрицательные).
4. `Linear(128, 10)` — выходной слой: 10 чисел, по одному на каждую цифру (0–9).

**Вход** сети: тензор `[батч, 1, 28, 28]`.
**Выход** сети: тензор `[батч, 10]` — «баллы» (логиты) для каждой из 10 цифр.
Чем больше балл, тем увереннее сеть в этой цифре.


```python
class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()          # [B,1,28,28] -> [B,784]
        self.fc1 = nn.Linear(28 * 28, 128)   # первый слой
        self.fc2 = nn.Linear(128, 10)        # выходной слой (10 цифр)

    def forward(self, x):
        x = self.flatten(x)      # разворачиваем картинку в вектор
        x = F.relu(self.fc1(x))  # первый слой + активация ReLU
        x = self.fc2(x)          # выходной слой (без активации — вернём логиты)
        return x

model = SimpleNet().to(device)   # переносим модель на выбранное устройство
print(model)
```

## Шаг 8. Функция потерь и оптимизатор

- **Функция потерь** (`CrossEntropyLoss`) измеряет, насколько ответ сети далёк
  от правильного. Чем меньше — тем лучше. Для задач классификации это стандарт.
- **Оптимизатор** (`Adam`) — алгоритм, который подкручивает веса сети, чтобы
  потери уменьшались. `lr` (learning rate) — размер шага обучения.


```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

## Шаг 9. Тренировочный цикл (само обучение)

Одна **эпоха** = один полный проход по всем обучающим данным.
Внутри эпохи для каждого батча повторяем 5 шагов:

1. **Прямой проход**: подаём картинки в сеть, получаем предсказания.
2. **Считаем потери**: насколько предсказания далеки от правды.
3. `optimizer.zero_grad()` — обнуляем старые градиенты.
4. `loss.backward()` — считаем градиенты (в какую сторону крутить веса).
5. `optimizer.step()` — делаем шаг: обновляем веса.

Обучим на 3 эпохи — этого хватит для хорошего результата.


```python
epochs = 3

for epoch in range(epochs):
    model.train()               # переводим модель в режим обучения
    running_loss = 0.0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        # 1. Прямой проход
        outputs = model(images)
        # 2. Потери
        loss = loss_fn(outputs, labels)

        # 3-5. Обратное распространение и шаг оптимизации
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    avg_loss = running_loss / len(train_loader)
    print(f"Эпоха {epoch + 1}/{epochs}  —  средние потери: {avg_loss:.4f}")

print("Готово! Сеть обучена.")
```

## Шаг 10. Проверяем точность на тестовых данных

Тестовые картинки сеть **не видела** во время обучения — поэтому по ним
честно оценивать качество.

`torch.no_grad()` отключает подсчёт градиентов (нам они тут не нужны,
так быстрее). `argmax` берёт индекс наибольшего балла — это и есть
предсказанная цифра.


```python
model.eval()   # режим оценки
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        predicted = outputs.argmax(dim=1)   # выбираем цифру с наибольшим баллом
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f"Точность на тесте: {accuracy:.2f}%  ({correct} из {total})")
```

## Шаг 11. Демонстрация: предсказание на конкретных картинках

Возьмём несколько тестовых картинок, покажем их и подпишем:
- **Ответ сети** (predicted)
- **Правильный ответ** (true)

Зелёный заголовок — сеть угадала, красный — ошиблась.


```python
model.eval()

# Берём один батч тестовых картинок
images, labels = next(iter(test_loader))
images_dev = images.to(device)

with torch.no_grad():
    outputs = model(images_dev)
    preds = outputs.argmax(dim=1).cpu()

# Рисуем первые 10 картинок
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i, ax in enumerate(axes.flat):
    ax.imshow(images[i].squeeze(), cmap="gray")
    pred = preds[i].item()
    true = labels[i].item()
    color = "green" if pred == true else "red"
    ax.set_title(f"Ответ: {pred}\nВерно: {true}", color=color)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Шаг 12. Как сеть «думает» над одной картинкой

Посмотрим на баллы для всех 10 цифр по одной картинке.
Через `softmax` превратим баллы в **вероятности** (в сумме дают 100%).
Самый высокий столбик — ответ сети.


```python
idx = 25   # можете поменять на любой индекс от 0 до 63

one_image = images[idx:idx+1].to(device)   # форма [1, 1, 28, 28]

with torch.no_grad():
    logits = model(one_image)
    probs = F.softmax(logits, dim=1).cpu().squeeze()

# Слева картинка, справа вероятности по цифрам
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.imshow(images[idx].squeeze(), cmap="gray")
ax1.set_title(f"Настоящая цифра: {labels[idx].item()}")
ax1.axis("off")

ax2.bar(range(10), probs.numpy())
ax2.set_xticks(range(10))
ax2.set_xlabel("Цифра")
ax2.set_ylabel("Вероятность")
ax2.set_title(f"Сеть уверена в цифре {probs.argmax().item()} "
              f"на {100 * probs.max().item():.1f}%")

plt.tight_layout()
plt.show()
```
